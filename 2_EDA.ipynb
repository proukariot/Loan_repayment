# Exploratory Data Analysis

```python
# import libraries
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
# load cleaned data
eda_data = pd.read_csv('c_data.csv')
eda_data = eda_data.drop(eda_data.columns[0], axis=1)
eda_data.head()
```

```python
#1 What is the correlation between fica score and loan status?
fig, ax = plt.subplots(figsize=(25, 5))
sns.countplot(data=eda_data, x='fico_range', hue = 'loan_status', palette= 'Set2', ax=ax)
```

### The percentage of loans that are unpaid increases as the FICO score decreases. For low FICO values, the ratio of loans that are paid to unpaid is smaller, compared to higher FICO scores.

```python
#2 How does credit age relate to the likelihood of default, and is this risk independent of or related to the FICO score?
fig, ax = plt.subplots(figsize=(5, 5))
sns.countplot(data=eda_data, x='credit_month', hue = 'loan_status', palette= 'Set2', ax=ax)
```

### A lower credit age indicates a higher chance of repaying the loan as compared to a higher age.

```python
sns.boxplot(x = eda_data['credit_month'],
            y = eda_data['fico_range'],
            hue = eda_data['loan_status'])
```

```python
fig, ax = plt.subplots(figsize=(25, 5))
sns.countplot(data=eda_data, x='fico_range', hue = 'credit_month', palette= 'Set2', ax=ax)
```

### The ratio of repaid to defaulted loans for the entire range of FICO scores is similar, indicating that FICO score is not associated with credit age.

```python
#3 How does home mortgage status relate to the likelihood of default?
fig, ax = plt.subplots(figsize=(5, 5))
sns.countplot(data=eda_data, x='home_ownership', hue = 'loan_status', palette= 'Set2', ax=ax)
```

### The ratio of repaid to outstanding loans for a mortgage is similar to that for a rental loan and is more likely to be repaid than a home equity loan.

```python
#4. How does annual income relate to default probability?
sns.boxplot(x = eda_data['loan_status'],
            y = eda_data[eda_data['annual_inc']<600000]['annual_inc'])
```

### Annual income has little significance in repaying the loan, people earning more per year have a slight advantage in repaying the loan.

```python
#5. How does employment history relate to the likelihood of default?
sns.boxplot(x = eda_data['loan_status'],
            y = eda_data['emp_length'])
```

### The average age for people who defaulted on their loans is greater than that of those who repaid them.

```python
fig, ax = plt.subplots(figsize=(5, 5))
sns.countplot(data=eda_data, x='emp_length', hue = 'loan_status', palette= 'Set2', ax=ax)
```

```python
#6. How is the size of the requested loan related to the probability of default?
sns.boxplot(x = eda_data['loan_status'],
            y = eda_data['loan_amnt'])
```

### The size of the loan has little impact on the likelihood of repaying it.
